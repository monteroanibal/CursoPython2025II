## Declaración y exploración de series y dataframes

A continuación, una explicación sobre cómo declarar o crear *Series* y *DataFrames* en *pandas*:

### Series

Una *Serie* es un arreglo unidimensional con etiquetas (índices) que pueden asociarse a cada valor. Es similar a una columna en una tabla de datos. A continuación se muestran las formas de creación de una Serie:

* **A partir de una lista o arreglo de Python**:

````python
import pandas as pd  

s = pd.Series([10, 20, 30, 40])  
print(s)
````

* **Especificando índices personalizados**:

````python
s = pd.Series([10, 20, 30, 40], index=["a", "b", "c", "d"])  
print(s)
````

* **A partir de un diccionario**:

````python
s = pd.Series({"x": 100, "y": 200, "z": 300})  
print(s)
````

### DataFrame

Un *DataFrame* es una estructura bidimensional que puede entenderse como una tabla de datos con filas y columnas. Cada columna es una *Serie* y todas comparten el mismo índice de filas. A continuación se muestran las formas de creación de un DataFrame:

* **A partir de un diccionario de listas o arreglos**: Las claves del diccionario se convierten en nombres de columna:

````python
data = {
    "Nombre": ["Ana", "Luis", "Marta"],
    "Edad": [23, 34, 29],
    "Ciudad": ["Bogotá", "Medellín", "Cali"]
}
df = pd.DataFrame(data)  
print(df)
````

* **A partir de una lista de diccionarios**: Cada diccionario representa una fila de datos:

````python
data = [
    {"Nombre": "Ana", "Edad": 23, "Ciudad": "Bogotá"},
    {"Nombre": "Luis", "Edad": 34, "Ciudad": "Medellín"},
    {"Nombre": "Marta", "Edad": 29, "Ciudad": "Cali"}
]
df = pd.DataFrame(data)  
print(df)
````

* **A partir de un arreglo de NumPy**: Se pueden combinar con etiquetas de columnas y filas:

````python
import numpy as np  

arr = np.array([[1, 2, 3], [4, 5, 6]])  
df = pd.DataFrame(arr, columns=["A", "B", "C"])  
print(df)
````

* **Creación vacía y llenado posterior**

````python
df = pd.DataFrame(columns=["Nombre", "Edad", "Ciudad"])  
print(df)
````

En resumen, *Series* representan datos unidimensionales con índices, mientras que los *DataFrames* permiten trabajar con información tabular de múltiples dimensiones. Estas estructuras constituyen la base para casi todas las operaciones de manipulación y análisis en *pandas*.

## Indexación y slicing

Una vez creadas las estructuras de datos en *pandas* (Series y DataFrames), es fundamental aprender a acceder y seleccionar la información que contienen. Esto se logra mediante indexación y *slicing*.

### Indexación en *Series*

Una *Serie* tiene un índice asociado a cada valor, lo que permite acceder a los elementos de manera similar a los diccionarios o listas de Python.

* **Acceso por posición (entero)**

````python
import pandas as pd  

s = pd.Series([10, 20, 30, 40], index=["a", "b", "c", "d"])  
print(s[0])      # Accede al primer elemento -> 10
````

* **Acceso por etiqueta (índice nombrado)**

````python
print(s["c"])    # Accede al elemento con índice "c" -> 30
````

* **Acceso a varios elementos**

````python
print(s[["a", "d"]])   # Devuelve una Serie con los índices "a" y "d"
````

### *Slicing* en *Series*

Permite obtener subconjuntos de datos, ya sea por posición o por etiquetas.

* **Por posición (estilo listas de Python)**

````python
print(s[1:3])   # Del índice 1 al 2 (excluye el 3)
````

* **Por etiquetas (incluye el final)**

````python
print(s["b":"d"])   # Incluye desde "b" hasta "d"
````

### Indexación en *DataFrames*

En los *DataFrames**, la selección puede hacerse por *filas* y/o *columnas*. Para ello, *pandas* ofrece principalmente dos métodos:

* **.loc[] → acceso por etiquetas**: Selección de una fila por índice:

````python
df = pd.DataFrame({
    "Nombre": ["Ana", "Luis", "Marta"],
    "Edad": [23, 34, 29],
    "Ciudad": ["Bogotá", "Medellín", "Cali"]
})

print(df.loc[0])        # Fila con índice 0
print(df.loc[:, "Edad"]) # Toda la columna "Edad"
````

* **.loc[] → acceso por etiquetas**: Selección de un rango de filas y columnas:

````python
print(df.loc[0:1, ["Nombre", "Edad"]]) 
````

* **.iloc[]` → acceso por posición numérica**: Selección de una fila:

````python
print(df.iloc[0])        # Primera fila
print(df.iloc[:, 1])     # Segunda columna (Edad)
````

* **.iloc[]` → acceso por posición numérica**: Selección de un rango:

````python
print(df.iloc[0:2, 0:2]) # Primeras dos filas y dos columnas
````

* **`loc` incluye el índice final, mientras que `iloc` lo excluye**, igual que en las listas de Python. Es posible combinar condiciones para filtrar datos:

````python
print(df.loc[df["Edad"] > 25, ["Nombre", "Ciudad"]])
````

## Eliminación de filas y columnas

En el trabajo con datos, es frecuente que se requiera descartar filas o columnas que no son necesarias para el análisis. La librería *pandas* facilita esta tarea a través del método *`drop()`*, entre otras opciones.

### Eliminación de *filas*

Por defecto, el método `drop()` elimina *filas* si se especifica su índice.

````python
import pandas as pd  

df = pd.DataFrame({
    "Nombre": ["Ana", "Luis", "Marta"],
    "Edad": [23, 34, 29],
    "Ciudad": ["Bogotá", "Medellín", "Cali"]
})

# Eliminar la fila con índice 1
df_filas = df.drop(1)
print(df_filas)
````

El DataFrame original no se modifica a menos que se use el parámetro `inplace=True`:

````python
df.drop(1, inplace=True)
````

### Eliminación de *columnas*

Para eliminar columnas se utiliza el parámetro `axis=1`.

````python
# Eliminar la columna "Ciudad"
df_columnas = df.drop("Ciudad", axis=1)
print(df_columnas)
````

También se pueden eliminar varias columnas a la vez:

````python
df_varias = df.drop(["Edad", "Ciudad"], axis=1)
````

### Eliminación condicional de filas

Es posible eliminar filas según una condición lógica. Por ejemplo, eliminar personas menores de 30 años:

````python
df_cond = df.drop(df[df["Edad"] < 30].index)
print(df_cond)
````

### Eliminación de valores nulos con dropna()

Además de `drop()`, pandas cuenta con métodos especializados para limpieza:

* **Eliminar filas con valores nulos**

````python
df_sin_nulos = df.dropna()
````

* **Eliminar columnas con valores nulos**

````python
df_sin_nulos_col = df.dropna(axis=1)
````

## Agrupación de filas

En análisis de datos, muchas veces no basta con tener la información fila por fila: necesitamos *resumirla*, *clasificarla* u *obtener estadísticas* a partir de grupos. Para ello, *pandas* ofrece las operaciones de *agrupación* (*grouping*) y *agregación* (*aggregation*).

### Agrupación con groupby()

El método **`groupby()`** permite dividir un DataFrame en grupos basados en los valores de una o más columnas. Ejemplo:

````python
import pandas as pd  

df = pd.DataFrame({
    "Ciudad": ["Bogotá", "Bogotá", "Medellín", "Cali", "Cali"],
    "Ventas": [100, 200, 150, 120, 180],
    "Vendedor": ["Ana", "Luis", "Ana", "Marta", "Luis"]
})

# Agrupar por ciudad
grupo = df.groupby("Ciudad")
print(grupo)
````

El resultado de `groupby()` no es un DataFrame directo, sino un objeto agrupado que se puede combinar con operaciones posteriores.

### Agregación de datos

Una vez definidos los grupos, se pueden aplicar funciones de agregación para obtener estadísticas:

* **Funciones comunes** (`sum`, `mean`, `count`, `max`, `min`):

````python
# Suma de ventas por ciudad
print(df.groupby("Ciudad")["Ventas"].sum())

# Promedio de ventas por ciudad
print(df.groupby("Ciudad")["Ventas"].mean())
````

* **Varias funciones a la vez**:

````python
print(df.groupby("Ciudad")["Ventas"].agg(["sum", "mean", "max"]))
````

### Agrupación por múltiples columnas

También es posible agrupar usando más de una columna:

````python
# Agrupar por Ciudad y Vendedor
print(df.groupby(["Ciudad", "Vendedor"])["Ventas"].sum())
````

Esto genera una tabla con índice jerárquico (MultiIndex).

### Uso de agg() para personalizar agregaciones

`agg()` permite aplicar diferentes funciones a distintas columnas:

````python
resumen = df.groupby("Ciudad").agg({
    "Ventas": ["sum", "mean"],      # sumar y promediar ventas
    "Vendedor": "count"             # contar vendedores
})
print(resumen)
````

### Reiniciar el índice después de agrupar

Por defecto, la columna usada para agrupar se convierte en índice. Para devolverla como columna normal:

````python
resultado = df.groupby("Ciudad")["Ventas"].sum().reset_index()
print(resultado)
````

## Concatenación y merge

En el análisis de datos, es común que la información se encuentre distribuida en diferentes tablas o estructuras. Para trabajar con ella de manera integrada, *pandas* ofrece mecanismos para *combinar* datos: la *concatenación* y el *merge*.

### Concatenación con pd.concat()

La función **`concat()`** permite unir objetos de *pandas* (Series o DataFrames) de manera vertical (uno debajo del otro) u horizontal (uno al lado del otro).

* **Concatenación vertical (apilar filas)**

````python
import pandas as pd  

df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Nombre": ["Ana", "Luis", "Marta"]
})

df2 = pd.DataFrame({
    "ID": [4, 5],
    "Nombre": ["Carlos", "Elena"]
})

df_concat = pd.concat([df1, df2], axis=0)
print(df_concat)
````

Se agregan las filas de `df2` debajo de las de `df1`.

* **Concatenación horizontal (unir columnas)**

````python
df3 = pd.DataFrame({
    "Edad": [23, 34, 29],
    "Ciudad": ["Bogotá", "Medellín", "Cali"]
})

df_horiz = pd.concat([df1, df3], axis=1)
print(df_horiz)
````

Aquí, las columnas se pegan una al lado de la otra, alineadas por el índice.

### *Merge* con pd.merge()

El método **`merge()`** permite combinar DataFrames de manera similar a un *join* en bases de datos relacionales. Se basa en claves comunes (columnas).

* **Inner join (intersección)**: Devuelve solo las filas con ID en común en ambos DataFrames.

````python
df_ventas = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Ventas": [100, 200, 150, 300]
})

df_clientes = pd.DataFrame({
    "ID": [1, 2, 3],
    "Nombre": ["Ana", "Luis", "Marta"]
})

df_inner = pd.merge(df_clientes, df_ventas, on="ID", how="inner")
print(df_inner)
````

* **Left join (todos los de la izquierda)**: Devuelve todos los clientes, aunque no tengan ventas asociadas.

````python
df_left = pd.merge(df_clientes, df_ventas, on="ID", how="left")
print(df_left)
````

* **Right join (todos los de la derecha)**:

````python
df_right = pd.merge(df_clientes, df_ventas, on="ID", how="right")
print(df_right)
````

* **Outer join (unión completa)**

````python
df_outer = pd.merge(df_clientes, df_ventas, on="ID", how="outer")
print(df_outer)
````

Devuelve todos los registros, incluso si no coinciden en ambos DataFrames (rellenando con `NaN` donde no hay correspondencia).

## Reshape y de tabla de contingencia

Cuando se trabaja con conjuntos de datos, muchas veces se necesita resumir la información en forma de tabla, calculando totales o promedios de manera agrupada. Para esto, *pandas* ofrece dos funciones poderosas: *`pivot_table()`* y *`crosstab()`*.

La función *`pivot_table()`* permite crear tablas dinámicas similares a las de Excel. Se utiliza para reorganizar y resumir datos a partir de un DataFrame. Sintaxis básica:

````python
pd.pivot_table(data, values, index, columns, aggfunc)
````

*data* → DataFrame de origen. *values* → columna(s) numéricas a resumir. *index* → filas de la tabla. *columns* → columnas de la tabla. *aggfunc* → función de agregación (ej: sum, mean, count).

Esto muestra el total de ventas por *Ciudad* (filas) y *Vendedor* (columnas).

````python
import pandas as pd  

df = pd.DataFrame({
    "Ciudad": ["Bogotá", "Bogotá", "Medellín", "Cali", "Cali"],
    "Vendedor": ["Ana", "Luis", "Ana", "Marta", "Luis"],
    "Ventas": [100, 200, 150, 120, 180]
})

tabla = pd.pivot_table(df, values="Ventas",
                       index="Ciudad",
                       columns="Vendedor",
                       aggfunc="sum",
                       fill_value=0)
print(tabla)
````

La función *`crosstab()`* crea una tabla de contingencia que muestra *frecuencias* (conteo de ocurrencias). Es muy útil para análisis de variables categóricas.

````python
pd.crosstab(index, columns, values=None, aggfunc=None)
````

*index* → categorías en filas. *columns* → categorías en columnas. *values + aggfunc* (opcional) → para resumir valores numéricos en lugar de contar. Por ejemplo aquí, se muestra cuántas veces aparece cada vendedor en cada ciudad.

````python
# Conteo de vendedores por ciudad
tabla_frec = pd.crosstab(df["Ciudad"], df["Vendedor"])
print(tabla_frec)
````

Con valores y función de agregación:

````python
# Suma de ventas por ciudad y vendedor
tabla_sum = pd.crosstab(df["Ciudad"], df["Vendedor"],
                        values=df["Ventas"],
                        aggfunc="sum")
print(tabla_sum)
````
