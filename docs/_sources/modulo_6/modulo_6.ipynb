## Análisis del Problema

  * **¿Qué es un mapa?**: Los mapas son representaciones gráficas a partir de medidas longitudinales, dentro del mismo se especifican rutas para establecer destinos de un punto a otro, allí se ubican localidades y nombres de las ciudades más importantes. A medida que avanzan las tecnologías, el diseño de los mapas se convierte en algo más sencillo; en la actualidad, por ejemplo, la mayoría de los mapas se realizan a partir de  fotografías aéreas y satelitales, lo cual permite llegar a una mayor precisión en la elaboración de los mismos. 

```{admonition} Nota
:class: tip
La tecnología para la creación de plataformas digitales actualmente conduce la generación de las técnicas y las pautas para la construcción de mapas.
```

## Funcionalidades de un mapa interactivo

 * **¿Actualmente, cuáles funcionalidades debería tener un mapa interactivo?**: Las funcionalidades en el diseño no son declaradas directamente por estos aprovisionadores de la solución para la construcción de mapas, pero se pueden rastrear los siguientes:
   - **_Ampliar_**: Permitir al usuario del mapa acercarse a zonas de interés y mover el mapa para explorar diferentes áreas.
   - **_Desplazar_**: Permitir al usuario mover el mapa para explorar diferentes áreas.
   - **_Rotar_**: En los casos en que la representación es en tres dimensiones, permitir al usuario rotal el mapa, tanto vertical como horizontalmente.
   - **_Ventanas emergentes tooltips_**: Al hacer clic o desplazar el cursor a un punto o elemento del mapa, se abre una ventana emergente con información adicional, como textos, imágenes, videos o enlaces. 
   - **_Activar y desactivar_**: Permitir al usuario del mapa poner y quitar capas de datos. Ejemplos de esto es el control de la opacidad oel _split_ para la superposición de capas.
   - **_Representación de tipos capas de información_**: Tanto ráster como vectorial.
   - **_filtros cruzados_**: resaltar elementos gráficos en diferentes figuras representadas en el mapa al navegar sobre los gráficos estadísticos con el cursor.
   - **_Medición_**: Brindarle al usiario las herramientas gráficas sobre el mapa para medir distancias, áreas. 
Guías y rutas:
   - **_Sistemas de navegación_**: para obtener indicaciones y rutas a lugares específicos. 
   - **_Actualización en tiempo real_**: la información sobre fenómenos como el tráfico, el clima y la población proyectada es suceptible a ser recuperada en tiempo real a manera de tablero de control.
   - **_Compartir y colaborar_**: Permitir al usuario compartir e interactuar conjuntamente con otros usuarios.
