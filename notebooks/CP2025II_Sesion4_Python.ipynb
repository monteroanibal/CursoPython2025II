# 1. Crear series

Pandas viene instalado con la instalación predeterminada de Python.

```python
import pandas as pd
import numpy as np
```

Una serie de Pandas es una estructura de datos que almacena datos en forma de columna. Normalmente, una serie se utiliza para almacenar información sobre un atributo específico del conjunto de datos. Veamos cómo crear una serie en Pandas.

Hay diferentes maneras de crear una serie con Pandas. El siguiente script importa el módulo Pandas y luego llama al constructor de la clase Series() para crear una serie vacía. Así es como se hace:

```python
my_series = pd.Series()
print(my_series)
```

También puedes crear una serie usando un array de NumPy. Pero primero, debes pasar el array al constructor de la clase Series():

```python
my_array = np.array([10, 20, 30, 40, 50])
my_series = pd.Series(my_array)
print(my_series)
```

En el resultado anterior, puede ver que los índices de una serie comienzan de 0 a 1 menos que el número de elementos de la serie.
También puede definir índices personalizados para su serie. Para ello, debe pasar su lista de índices al atributo index de la clase Series, como se muestra en el siguiente script:

```python
my_array = np.array([10, 20, 30, 40, 50])
my_series = pd.Series(my_array, index = ["num1", "num2", "num3", "num4","num5"])
print(my_series)
```

También puedes crear una serie pasando directamente una lista de Python al constructor de la clase Series().

```python
my_series = pd.Series([10, 20, 30, 40, 50], index = ["num1", "num2","num3", "num4", "num5"])
print(my_series)
```

Finalmente, un valor de escalador también se puede usar para definir una serie. Si se pasa una lista de índices, el valor de escalador se repetirá tantas veces como elementos de la lista de índices. A continuación, se muestra un ejemplo:

```python
my_series = pd.Series(25, index = ["num1", "num2", "num3", "num4","num5"])
print(my_series)
```

Finalmente, también puede crear una serie usando un diccionario. En este caso, las claves del diccionario se convertirán en índices de la serie, mientras que los valores del diccionario se insertarán como elementos de la serie. A continuación, se muestra un ejemplo:

```python
my_dict = {
'num1':6,
'num2':7,
'num3':8
}
my_series = pd.Series(my_dict)
print(my_series)
```

Veamos algunas de las operaciones útiles que puedes realizar con la serie Pandas. Puedes usar corchetes y etiquetas de índice para acceder a los elementos de la serie, como se muestra en el siguiente script:

```python
my_series = pd.Series([10, 20, 30, 40, 50], index = ["num1", "num2","num3", "num4", "num5"])
print(my_series[0])
print(my_series['num3'])
```

Usando las funciones min() y max() del módulo NumPy, puedes encontrar los valores máximo y mínimo, respectivamente, de una serie. Consulta el siguiente script como referencia.

```python
my_series = pd.Series([5, 8, 2, 11, 9])
print(np.min(my_series))
print(np.max(my_series))
```

De manera similar, el método mean() del módulo NumPy puede encontrar la media de una serie de Pandas, como se muestra en el siguiente script.

```python
my_series = pd.Series([5, 8, 2, 11, 9])
print(my_series.mean())
```

El siguiente script encuentra el valor mediano de una serie de Pandas.

```python
my_series = pd.Series([5, 8, 2, 11, 9])
print(my_series.median())
```

También puedes encontrar el tipo de dato de una serie de Pandas usando el atributo dtype. Aquí tienes un ejemplo:

```python
my_series = pd.Series([5, 8, 2, 11, 9])
print(my_series.dtype)
```

Una serie de Pandas también se puede convertir en una lista de Python usando el método tolist(), como se muestra en el siguiente script:

```python
my_series = pd.Series([5, 8, 2, 11, 9])
print(my_series.tolist())
```

# 2. Crear dataframes

Un dataframe de Pandas es una estructura de datos tabular que almacena datos en forma de filas y columnas. Por norma general, las filas corresponden a registros, mientras que las columnas a atributos. En pocas palabras, un dataframe de Pandas es una colección de series.

Al igual que con una serie, existen varias maneras de crear un dataframe de Pandas. Para crear un dataframe vacío, puede usar la clase DataFrame del módulo Pandas, como se muestra a continuación:

```python
my_df = pd.DataFrame()
print(my_df)
```

Puedes crear un dataframe de Pandas usando una lista de listas. Cada sublista de la lista externa corresponde a una fila de un dataframe.
Cada elemento de una sublista se convierte en un valor de atributo.
Para especificar encabezados de columna, debes pasar una lista de valores al atributo "columns" de la clase DataFrame.
Aquí tienes un ejemplo de cómo crear un dataframe de Pandas usando una lista.

```python
scores = [['Mathematics', 85], ['English', 91], ['History', 95]]
my_df = pd.DataFrame(scores, columns = ['Subject', 'Score'])
my_df
```

De forma similar, puedes crear un dataframe de Pandas usando un diccionario. Una forma de hacerlo es crear un diccionario donde las claves corresponden a los encabezados de columna. Por otro lado, los valores correspondientes del diccionario son una lista, que corresponde a los valores de columna en el dataframe de Pandas.
Aquí tienes un ejemplo como referencia:

```python
scores = {'Subject':["Mathematics", "History", "English", "Science","Arts"],
'Score':[98, 75, 68, 82, 99]
}
my_df = pd.DataFrame(scores)
my_df
```

Otra forma de crear un dataframe de Pandas es usar una lista de diccionarios. Cada diccionario corresponde a una fila. Aquí tienes un ejemplo de cómo hacerlo.

```python
scores = [
{'Subject':'Mathematics', 'Score':85},
{'Subject':'History', 'Score':98},
{'Subject':'English', 'Score':76},
{'Subject':'Science', 'Score':72},
{'Subject':'Arts', 'Score':95},
]
my_df = pd.DataFrame(scores)
my_df
```

Los diccionarios de la lista utilizados para crear un dataframe de Pandas no necesitan tener el mismo tamaño.
Por ejemplo, en el script a continuación, el cuarto diccionario de la lista contiene solo un elemento, a diferencia del resto de los diccionarios de esta lista. El dataframe correspondiente contendrá un valor nulo en lugar del segundo elemento, como se muestra en el resultado del script a continuación:

```python
scores = [
{'Subject':'Mathematics', 'Score':85},
{'Subject':'History', 'Score':98},
{'Subject':'English', 'Score':76},
{'Score':72},
{'Subject':'Arts', 'Score':95},
]
my_df = pd.DataFrame(scores)
my_df
```

Veamos ahora algunas de las operaciones básicas que puedes realizar en los dataframes de Pandas.
Para ver las primeras (N) filas de un dataframe, puedes llamar al método head(), como se muestra en el siguiente script:

```python
scores = [
{'Subject':'Mathematics', 'Score':85},
{'Subject':'History', 'Score':98},
{'Subject':'English', 'Score':76},
{'Subject':'Science', 'Score':72},
{'Subject':'Arts', 'Score':95},
]
my_df = pd.DataFrame(scores)
my_df.head(2)
```

Para ver las últimas N filas, puedes usar el método tail(). Aquí tienes un ejemplo:

```python
my_df = pd.DataFrame(scores)
my_df.tail(2)
```

También puedes obtener un resumen de tu marco de datos de Pandas usando el método info().

```python
my_df = pd.DataFrame(scores)
my_df.info()
```

En el resultado anterior, puede ver la cantidad de entradas en su marco de datos de Pandas, la cantidad de columnas junto con su tipo, etc.
Finalmente, para obtener información como la media, el mínimo, el máximo, la desviación estándar, etc., de las columnas numéricas en el dataframe de Pandas, puedes usar el método describe(), como se muestra en el siguiente script:

```python
my_df = pd.DataFrame(scores)
my_df.describe()
```

```python
type(titanic_data["class"])
```

# 6. Eliminando filas y columnas

El siguiente script crea un marco de datos ficticio que utilizarás en esta sección.

```python
scores = [
{'Subject':'Mathematics', 'Score':85, 'Grade': 'B', 'Remarks': 'Good',},
{'Subject':'History', 'Score':98, 'Grade': 'A','Remarks':'Excellent'},
{'Subject':'English', 'Score':76, 'Grade': 'C','Remarks': 'Fair'},
{'Subject':'Science', 'Score':72, 'Grade': 'C','Remarks': 'Fair'},
{'Subject':'Arts', 'Score':95, 'Grade': 'A','Remarks': 'Excellent'},
]
my_df = pd.DataFrame(scores)
my_df.head()
```

El siguiente script elimina registros en los índices 1 y 4.

```python
my_df2 = my_df.drop([1,4])
my_df2.head()
```

En el resultado anterior, puede ver que los índices no están en secuencia, ya que ha eliminado los índices 1 y 4.
Puede restablecer los índices del dataframe desde 0 mediante el método reset_index().
Invoquemos el método reset_index() en el dataframe my_df2.
Aquí, el valor "True" del parámetro "inplace" especifica que desea eliminar los registros in situ sin asignar el resultado a ninguna variable nueva.

```python
my_df2.reset_index(inplace=True)
my_df2.head()
```

El resultado anterior muestra que los índices se han restablecido.
También puede ver que se ha añadido un nuevo índice de columna, que contiene el índice original. Si solo desea restablecer los nuevos índices sin crear una nueva columna llamada index, puede hacerlo pasando "True" como valor para el parámetro "drop" del método reset_index.
De nuevo, eliminemos algunas filas y restablezcamos el índice mediante el método reset_index(), pasando "True" como valor para el atributo "drop". Vea los dos scripts siguientes:

```python
my_df2 = my_df.drop([1,4])
my_df2.head()
```

# 9. Merge

Puedes fusionar varios marcos de datos basándose en valores comunes entre cualquiera de sus columnas.
Creemos dos marcos de datos: puntuaciones1 y puntuaciones2.

```python
scores1 = [
{'Subject':'Mathematics', 'Score':85, 'Grade': 'B', 'Remarks': 'Good',},
{'Subject':'History', 'Score':98, 'Grade': 'A','Remarks':'Excellent'},
{'Subject':'English', 'Score':76, 'Grade': 'C','Remarks': 'Fair'},
{'Subject':'Chemistry', 'Score':72, 'Grade': 'C','Remarks': 'Fair'},
]
scores2 = [
{'Subject':'Arts', 'Score':70, 'Grade': 'C','Remarks': 'Fair'},
{'Subject':'Physics', 'Score':75, 'Grade': 'C','Remarks': 'Fair'},
{'Subject':'English', 'Score':92, 'Grade': 'A','Remarks':'Excellent'},
{'Subject':'Chemistry', 'Score':91, 'Grade': 'A','Remarks':'Excellent'},
]
scores1_df = pd.DataFrame(scores1)
scores2_df = pd.DataFrame(scores2)
```

La función merge() de Pandas permite fusionar dos dataframes basándose en valores comunes entre sus columnas. Merge() es similar a las operaciones JOIN de SQL.

```python
join_inner_df = scores1_df.merge(scores2_df, on='Subject', how='inner')
join_inner_df.head()
```

El siguiente script fusiona los dataframes "scores1" y "scores2" mediante la estrategia INNER JOIN.

Para fusionar dos dataframes, se llama a la función merge() en el dataframe que se desea colocar a la izquierda de la operación de unión. El dataframe que se fusionará a la derecha se pasa como primer parámetro. A continuación, se pasa el nombre de la columna a la que se desea aplicar la operación de fusión al atributo "on". Finalmente, la estrategia de fusión se pasa al atributo "how", como se muestra en el siguiente script.

Con INNER JOIN, solo se devuelven los registros de ambos dataframes que tienen valores comunes en la columna utilizada para fusionarlos.
Por ejemplo, en los dataframes "scores1" y "scores2", Inglés y Química son dos asignaturas que existen en la columna "Asignatura" de ambos dataframes.

Por lo tanto, en el resultado del siguiente script, se pueden ver los dataframes fusionados que contienen solo estos dos registros. Dado que los nombres de las columnas Puntuación, Calificación y Comentarios son similares en ambos marcos de datos, a los nombres de las columnas del conjunto de datos de la izquierda (que es puntuaciones1 en el siguiente script) se les ha añadido "_x". Mientras que a los nombres de las columnas del marco de datos de la derecha se les ha añadido "_y".

Si ambos marcos de datos en la operación de fusión contienen nombres de columna diferentes, verá los nombres de las columnas originales en el marco de datos fusionado.

```python
join_inner_df = scores1_df.merge(scores2_df, on='Subject', how='inner')
join_inner_df
```

Al fusionar dos marcos de datos de Pandas mediante una unión IZQUIERDA, se devuelven todos los registros del marco de datos situado a la izquierda de la función merge(), mientras que, para el marco de datos derecho, solo se devuelven aquellos registros donde se encuentra un valor común en la columna que se está fusionando.

En el resultado a continuación, puede ver los registros de las asignaturas de Matemáticas e Historia en el marco de datos de la izquierda. Sin embargo, como estos registros no existen en el marco de datos de la derecha, se ven valores nulos en las columnas Puntuación_y, Calificación_y y Comentarios_y. Solo se pueden ver en el marco de datos de la derecha aquellos registros donde hay valores comunes en la columna Asignatura de ambos marcos de datos.

```python
join_inner_df = scores1_df.merge(scores2_df, on='Subject', how='left')
join_inner_df.head()
```

En el resultado a continuación, puede ver los registros de las asignaturas de Matemáticas e Historia en el marco de datos de la izquierda. Sin embargo, como estos registros no existen en el marco de datos de la derecha, se ven valores nulos en las columnas Puntuación_y, Calificación_y y Comentarios_y. Solo se pueden ver en el marco de datos de la derecha aquellos registros donde hay valores comunes en la columna Asignatura de ambos marcos de datos.

```python
join_inner_df = scores1_df.merge(scores2_df, on='Subject', how='right')
join_inner_df.head()
```
